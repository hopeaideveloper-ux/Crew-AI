**CREW BASED STRING KNOWLEDGE SOURCE**

```python
from crewai import Agent, Task, Crew, Process
from crewai.knowledge.source.string_knowledge_source import StringKnowledgeSource

# Knowledge Source
policy_content = """
Company working hours are from 9:00 AM to 6:00 PM.
Employees receive 20 paid leave days per year.
Remote work is allowed two days per week.
"""

policy_source = StringKnowledgeSource(
    content=policy_content
)

# Agent
agent = Agent(
    role="HR Assistant",
    goal="Answer employee questions using company policy",
    backstory="You help employees understand company policies.",
    verbose=True
)

# Task
task = Task(
    description="Answer this employee question: {question}",
    expected_output="A simple answer based on the company policy.",
    agent=agent
)

# Crew
crew = Crew(
    agents=[agent],
    tasks=[task],
    process=Process.sequential,
    knowledge_sources=[policy_source],
    verbose=True
)

# Run
result = await crew.kickoff_async(inputs={"question": "How many paid leave days do employees get?"})
print("\nAnswer:")
print(result)
```

****AGENT ONLY KNOWLEDGE WITH CSV KNOWLEDGE SOURCE**

```python
from crewai import Agent, Task, Crew
from crewai.knowledge.source.csv_knowledge_source import CSVKnowledgeSource

# Agent-specific CSV knowledge
sales_knowledge = CSVKnowledgeSource(file_paths=["sales_dataset.csv"])

# Agent
sales_agent = Agent(
    role="Sales Analyst",
    goal="Answer questions using company sales data",
    backstory="You analyze sales records and provide simple business insights.",
    knowledge_sources=[sales_knowledge],
    verbose=True
)

# Task
task = Task(
    description="Answer this question using the sales data: {question}.",
    expected_output="A simple and accurate answer based on the CSV sales records.",
    agent=sales_agent
)

# No crew-level knowledge required
crew = Crew(
    agents=[sales_agent],
    tasks=[task]
)

# Run
result = await crew.kickoff_async(inputs={"question": "What is the total sales of Laptop?"})
print("\nAnswer:")
print(result)
```

**CREW WIDE KNOWLEDGE AND AGENT SPECIFIC KNOWLEDGE WITH TEXTFILE KNOWLEDGE SOURCE**

```python
from crewai import Agent, Task, Crew
from crewai.knowledge.source.text_file_knowledge_source import TextFileKnowledgeSource

# Crew-wide knowledge
crew_knowledge = TextFileKnowledgeSource(file_paths=["company_policy.txt"])

# Agent-specific knowledge
agent_knowledge = TextFileKnowledgeSource(file_paths=["technical_notes.txt"])

# Agent
agent = Agent(
    role="Technical Support Specialist",
    goal="Answer technical questions using company and technical knowledge",
    backstory="You are a technical support expert.",
    knowledge_sources=[agent_knowledge],   # Agent-specific knowledge
    verbose=True
)

# Task
task = Task(
    description="Answer this question: {question}",
    expected_output="A simple answer based on the available knowledge.",
    agent=agent
)

# Crew
crew = Crew(
    agents=[agent],
    tasks=[task],
    knowledge_sources=[crew_knowledge],   # Crew-wide knowledge
    verbose=True
)

# Run
result = await crew.kickoff_async(inputs={"question": "What are the company support hours and the server requirements?"})
print("\nAnswer:")
print(result)
```

**MULTIPLE AGENT WITH DIFFERENT KNOWLEDGE SOURCE**

```python
from crewai import Agent, Task, Crew
from crewai.knowledge.source.string_knowledge_source import StringKnowledgeSource


# Different knowledge for different agents
product_knowledge = StringKnowledgeSource(
    content="""
    Laptop Pro X costs ₹65,000.
    It has 16 GB RAM, 512 GB SSD, and an Intel Core i7 processor.
    """
)

shipping_knowledge = StringKnowledgeSource(
    content="""
    Standard delivery takes 3 to 5 business days.
    Express delivery takes 1 to 2 business days.
    Free shipping is available for orders above ₹5,000.
    """
)

return_knowledge = StringKnowledgeSource(
    content="""
    Products can be returned within 7 days of delivery.
    The product must be unused and in its original packaging.
    Refunds are processed within 5 business days after approval.
    """
)


# Agent 1
product_agent = Agent(
    role="Product Assistant",
    goal="Answer questions about product details and pricing",
    backstory="You help customers understand product specifications.",
    knowledge_sources=[product_knowledge],
    verbose=True
)


# Agent 2
shipping_agent = Agent(
    role="Shipping Assistant",
    goal="Answer questions about delivery and shipping",
    backstory="You help customers understand delivery options.",
    knowledge_sources=[shipping_knowledge],
    verbose=True
)


# Agent 3
return_agent = Agent(
    role="Return Assistant",
    goal="Answer questions about returns and refunds",
    backstory="You help customers understand return and refund policies.",
    knowledge_sources=[return_knowledge],
    verbose=True
)


# Tasks
product_task = Task(
    description="Answer this question about the product: {question}",
    expected_output="A clear answer about product details.",
    agent=product_agent
)

shipping_task = Task(
    description="Answer this question about shipping: {question}",
    expected_output="A clear answer about shipping.",
    agent=shipping_agent
)

return_task = Task(
    description="Answer this question about returns: {question}",
    expected_output="A clear answer about returns and refunds.",
    agent=return_agent
)


# Crew
crew = Crew(
    agents=[
        product_agent,
        shipping_agent,
        return_agent
    ],
    tasks=[
        product_task,
        shipping_task,
        return_task
    ],
    verbose=True
)


# Run
result = await crew.kickoff_async(
    inputs={
        "question": "What is the price of Laptop Pro X, how long will delivery take, and can I return it?"
    }
)

print("\nFinal Answer:")
print(result)
```
