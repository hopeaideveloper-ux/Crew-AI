**Flow - A CrewAI Flow is a way to control the steps of an application. You define where the workflow starts, which step runs next, and what information the steps share.**

**1. @start() - Identifies the first method to execute**

```python
from crewai.flow.flow import Flow, start
from pathlib import Path

class SimpleFlow(Flow):
    @start()
    def begin(self):
        message = "Simple Flow executed successfully"
        print(message)
        return message

flow = SimpleFlow()
flow.plot()

# Run the flow
result = flow.kickoff()
print("Final Result:", result)
```

**2. @listen() - Executes after another method finishes and receives its output.**

```python
from crewai.flow.flow import Flow, start, listen
from pathlib import Path

class ListenFlow(Flow):
    @start()
    def get_number(self):
        number = 10
        print("Starting Number:", number)
        return number
        
    @listen(get_number)
    def multiply_number(self, number):
        result = number * 2
        print("Multiplied Result:", result)
        return result

flow = ListenFlow()
flow.plot()
result = flow.kickoff()
print("Final Result:", result)
```

**3. @router - Selects the next execution path.**

```python
import random
from crewai.flow.flow import Flow, listen, router, start
from pydantic import BaseModel

class ExampleState(BaseModel):
    success_flag: bool = False

class RouterFlow(Flow[ExampleState]):

    @start()
    def start_method(self):
        print("Starting the structured flow")
        random_boolean = random.choice([True, False])
        self.state.success_flag = random_boolean

    @router(start_method)
    def second_method(self):
        if self.state.success_flag:
            return "success"
        else:
            return "failed"

    @listen("success")
    def third_method(self):
        print("Third method running")

    @listen("failed")
    def fourth_method(self):
        print("Fourth method running")


flow = RouterFlow()
flow.plot()
result = flow.kickoff()
print("Final Result:", result)
```

**4. STATE - SHARES INFORMATION --> State stores information that can be accessed by different Flow methods.**

```python
from crewai.flow.flow import Flow, start, listen
from pathlib import Path

class StateFlow(Flow):

    @start()
    def store_student(self):
        self.state["name"] = "John"
        self.state["score"] = 85

        print("Student information stored")
        return "Data stored"

    @listen(store_student)
    def display_student(self, previous_result):
        result = (
            f"Name: {self.state['name']}, "
            f"Score: {self.state['score']}"
        )

        print(result)
        return result

flow = StateFlow()
flow.plot()
result = flow.kickoff()
print("Final Result:", result)
print("Complete State:", dict(flow.state))
```

**5. Methods – Perform Operations --> Methods contain the actual operations performed by the Flow.**

```python
from crewai.flow.flow import Flow, start, listen
from pathlib import Path

class CalculationFlow(Flow):

    def add_numbers(self, first_number, second_number):
        return first_number + second_number

    @start()
    def calculate(self):
        result = self.add_numbers(10, 20)

        print("Addition Result:", result)
        return result

    @listen(calculate)
    def show_result(self, result):
        message = f"The calculated value is {result}"

        print(message)
        return message

flow = CalculationFlow()
flow.plot()
result = flow.kickoff()
print("Final Result:", result)
```

**6. Crews – Execute Agent-Based Work --> This example executes a Crew inside a Flow.**

```python
from crewai import Agent, Task, Crew, Process, LLM
from crewai.flow.flow import Flow, start, listen
from pathlib import Path

# LLM
llm = LLM(model="gpt-4o-mini", temperature=0)

# Agent
teacher = Agent(
    role="AI Teacher",
    goal="Explain AI concepts in simple language",
    backstory="You teach beginner-level AI students.",
    llm=llm,
    verbose=False
)

# Task
explanation_task = Task(
    description="Explain Artificial Intelligence in one short sentence.",
    expected_output="One beginner-friendly sentence.",
    agent=teacher
)

# Crew
teaching_crew = Crew(
    agents=[teacher],
    tasks=[explanation_task],
    process=Process.sequential,
    verbose=False
)

# Flow
class CrewFlow(Flow):

    @start()
    def begin(self):
        print("Starting the teaching Flow")
        return "Flow started"

    @listen(begin)
    def run_crew(self, previous_result):
        crew_result = teaching_crew.kickoff()

        answer = crew_result.raw
        print("Crew Answer:", answer)

        return answer

flow = CrewFlow()
flow.plot()
result = flow.kickoff()
print("Final Result:", result)
```

**7. Human in the Loop (human feedback) - The @human_feedback decorator enables human-in-the-loop workflows by pausing flow execution to collect feedback from a human.**

```python
from crewai.flow.flow import Flow, start, listen
from crewai.flow.human_feedback import human_feedback

class ReviewFlow(Flow):
    @start()
    @human_feedback(message="Please review this message:")
    def create_message(self):
        return "Artificial Intelligence helps automate tasks."

    @listen(create_message)
    def show_feedback(self, result):
        final_result = f"Human Feedback: {result.feedback}"
        print(final_result)
        return final_result

flow = ReviewFlow()
flow.plot()
result = flow.kickoff()
print("Final Result:", result)
```

**8. @human_feedback Routing with emit**

**When you don’t specify emit, the decorator simply collects feedback and passes a HumanFeedbackResult to the next listener**<br>
**When you specify emit, the decorator becomes a router. The human’s free-form feedback is interpreted by an LLM and collapsed into one of the specified outcomes**

```python
from crewai.flow.flow import Flow, start, listen
from crewai.flow.human_feedback import (human_feedback, HumanFeedbackResult)

class ApprovalFlow(Flow):

    @start()
    @human_feedback(
        message="Approve or reject this content:",
        emit=["approved", "rejected"],
        llm="gpt-4o-mini",
        default_outcome="rejected"
    )
    def create_content(self):
        return "AI can automate repetitive tasks."

    @listen("approved")
    def on_approved(self, result: HumanFeedbackResult):
        output = f"APPROVED: {result.feedback}"
        print(output)
        return output

    @listen("rejected")
    def on_rejected(self, result: HumanFeedbackResult):
        output = f"REJECTED: {result.feedback}"
        print(output)
        return output


flow = ApprovalFlow()
flow.plot()
result = flow.kickoff()
print("Final Result:", result)
```

**9. @human_feedback with Content Approval Workflow**

```python
from crewai.flow.flow import Flow, start, listen, or_
from crewai.flow.human_feedback import human_feedback, HumanFeedbackResult


class ContentApprovalFlow(Flow):

    @start()
    def create_draft(self):
        self.state["content"] = (
            "Artificial Intelligence helps automate repetitive tasks."
        )
        self.state["version"] = 0

        return self.state["content"]

    @human_feedback(
        message="Approve, reject, or request revision:",
        emit=["approved", "rejected", "revise"],
        llm="gpt-4o-mini",
        default_outcome="revise"
    )
    @listen(or_("create_draft", "revise"))
    def review_content(self):
        self.state["version"] += 1

        content = self.state["content"]

        print("\nContent:", content)
        print("Version:", self.state["version"])

        return content

    @listen("approved")
    def publish_content(self, result: HumanFeedbackResult):
        print("\nApproved Content:", result.output)
        print("Human Feedback:", result.feedback)

        return "Content published successfully"

    @listen("rejected")
    def discard_content(self, result: HumanFeedbackResult):
        print("\nRejected Content:", result.output)
        print("Reason:", result.feedback)

        return "Content rejected"


flow = ContentApprovalFlow()
flow.plot()
result = flow.kickoff()
print("\nFinal Result:", result)
```

**10. OR Conditional Execution - or_() runs the listener when any one of the specified methods finishes.**

```python
from crewai.flow.flow import Flow, start, listen, or_

class OrFlow(Flow):

    @start()
    def first_method(self):
        result = "First method completed"
        print(result)
        return result

    @listen(first_method)
    def second_method(self):
        result = "Second method completed"
        print(result)
        return result

    @listen(or_(first_method, second_method))
    def show_result(self, result):
        print("OR received:", result)

flow = OrFlow()
flow.plot()
flow.kickoff()
```

**11. AND Conditional Execution - and_() runs the listener only after all specified methods finish.**

```python
from crewai.flow.flow import Flow, start, listen, and_

class AndFlow(Flow):

    @start()
    def first_method(self):
        self.state["first"] = "First method completed"
        print(self.state["first"])

    @listen(first_method)
    def second_method(self):
        self.state["second"] = "Second method completed"
        print(self.state["second"])

    @listen(and_(first_method, second_method))
    def show_result(self):
        result = "Both methods completed"
        print("AND Result:", result)
        return result


flow = AndFlow()
flow.plot()
result = flow.kickoff()
print("Final Result:", result)
```

**12. ADDING AGENT TO A FLOW**

```python
from crewai import Agent
from crewai.flow.flow import Flow, start, listen

# Create Agent
teacher = Agent(
    role="AI Teacher",
    goal="Explain AI concepts simply",
    backstory="You teach AI to beginners.",
    llm="gpt-4o-mini",
    verbose=False
)

class AgentFlow(Flow):

    @start()
    def get_question(self):
        return "What is Artificial Intelligence?"

    @listen(get_question)
    def run_agent(self, question):
        result = teacher.kickoff(question)

        print("Agent Answer:")
        print(result.raw)

        return result.raw

flow = AgentFlow()
flow.plot()
result = flow.kickoff()
print("\nFinal Result:", result)
```

**13. Adding Crews to Flow - Creating a flow with multiple crews in CrewAI is straightforward.**

```python
from crewai import Agent, Task, Crew, Process
from crewai.flow.flow import Flow, start, listen
from pydantic import BaseModel

# Agent
writer = Agent(
    role="Blog Writer",
    goal="Write simple educational blog posts",
    backstory="You explain technical concepts to beginners.",
    llm="gpt-4o-mini",
    verbose=False
)

# Task
blog_task = Task(
    description="Write a short blog about {topic}.",
    expected_output="A beginner-friendly blog containing two paragraphs.",
    agent=writer
)

# Crew
blog_crew = Crew(
    agents=[writer],
    tasks=[blog_task],
    process=Process.sequential,
    verbose=False
)

# Flow State
class BlogState(BaseModel):
    topic: str = ""
    blog: str = ""

# Flow
class BlogFlow(Flow[BlogState]):

    @start()
    def prepare_topic(self):
        print("Blog Topic:", self.state.topic)
        return self.state.topic

    @listen(prepare_topic)
    def generate_blog(self, topic):
        result = blog_crew.kickoff(
            inputs={"topic": topic}
        )

        self.state.blog = result.raw
        print("\nGenerated Blog:\n")
        print(self.state.blog)
        return self.state.blog

    @listen(generate_blog)
    def save_blog(self, blog):
        with open("blog_post.md", "w", encoding="utf-8") as file:
            file.write(blog)
        print("\nBlog saved as blog_post.md")
        return blog

# User input
topic = input("Enter the blog topic: ")

flow = BlogFlow()
flow.plot()
result = flow.kickoff(inputs={"topic": topic})
print("\nFinal Result:\n", result)
```
