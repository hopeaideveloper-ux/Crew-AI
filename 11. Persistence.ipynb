**FLOW PERSISTENCE - Flow Persistence in CrewAI means saving a Flow’s state to storage so it can be restored later—even after the program stops or restarts.**

**Class-Level Persistence - Place @persist above the class to save state after each Flow method finishes.**

```python
from crewai.flow.flow import Flow, start, listen
from crewai.flow.persistence import persist
from pydantic import BaseModel

@persist()
class ClassFlow(Flow):

    @start()
    def first_step(self):
        self.state["counter"] = 1
        print("First step:", self.state["counter"])

    @listen(first_step)
    def second_step(self):
        self.state["counter"] += 1
        print("Second step:", self.state["counter"])

flow = ClassFlow()
flow.plot()
await flow.kickoff_async()
```

**Method-Level Persistence - Place @persist above one method to save the Flow’s state when that method finishes.**

```python
from crewai.flow.flow import Flow, start, listen
from crewai.flow.persistence import persist
from pydantic import BaseModel

class MethodFlow(Flow):

    @start()
    def first_step(self):
        self.state["counter"] = 1
        print("First step:", self.state["counter"])

    @persist()
    @listen(first_step)
    def save_step(self):
        self.state["counter"] += 1
        print("Saved counter:", self.state["counter"])

flow = MethodFlow()
flow.plot()
await flow.kickoff_async()
```

**Forking Persisted State - restore_from_state_id loads saved data into a new state ID, keeping the original history separate.**

```python
from crewai.flow.flow import Flow, start, listen
from crewai.flow.persistence import persist
from pydantic import BaseModel

class CounterState(BaseModel):
    counter: int = 0

@persist()
class CounterFlow(Flow[CounterState]):

    @start()
    def increase(self):
        self.state.counter += 1
        print("Counter:", self.state.counter)

original = CounterFlow()
await original.kickoff_async()

forked = CounterFlow()
await forked.kickoff_async(restore_from_state_id=original.state.id)
print("Different IDs:", original.state.id != forked.state.id)
```
