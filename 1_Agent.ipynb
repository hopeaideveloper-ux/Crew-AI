**CrewAI is an open-source Python framework for building multi-agent AI systems.**

**Difference between normal LLM, AI Agent, and Multi-Agent System**

| Type                   |  Explanation                                                                          |
| ---------------------- | ------------------------------------------------------------------------------------- |
| **Normal LLM**         | Takes a prompt and generates a response. It mainly answers questions or creates text. |
| **AI Agent**           | An LLM that can plan, make decisions, use tools, and perform tasks.                   |
| **Multi-Agent System** | Multiple AI agents work together, each handling a specific role or task.              |

**AGENTS - It is an AI worker with a specific role, goal, and responsibility.**

**Common Agent Properties**<br>
-role – what the agent is<br>
-goal – what the agent should achieve<br>
-backstory – background/context for the agent<br>
-llm – language model used by the agent<br>
-tools – external tools the agent can use<br>
-memory – lets the agent retain useful information<br>
-verbose – shows execution details<br>
-allow_delegation – allows the agent to delegate work to other agents<br>

```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()
llm = ChatOpenAI(
    model="gpt-5-nano",
    api_key=os.getenv("OPENAI_API_KEY"),
    temperature=1
)
```

**SINGLE AGENT EXAMPLE**

```python
from crewai import Agent

researcher = Agent(
    role="AI Researcher",
    goal="Explain Artificial Intelligence in simple words",
    backstory="You are an expert in AI and explain concepts clearly.",
    verbose=True
)

result = await researcher.kickoff("Explain Artificial Intelligence in simple words with one example.")
print(result)
```

**MANAGER AGENT WITH DELEGATION**

```python
from crewai import Agent

manager = Agent(
    role="Project Manager",
    goal="Manage tasks and coordinate team members",
    backstory="You are responsible for completing projects successfully.",
    allow_delegation=True,
    memory=True,
    verbose=False
)

result = await manager.kickoff( "Explain how a Project Manager manages tasks and coordinates team members.")
print(result)
```

**AGENT USING CUSTOM PROMPT TEMPLATE & LLM**

```python
from crewai import Agent

# Create an Agent
agent = Agent(
    role="AI Teacher",
    goal="Explain AI concepts clearly to beginners",
    backstory="You are an experienced AI teacher who explains technical topics simply.",
    llm=llm,
    verbose=True
)

# Get user question
question = input("Enter your question: ")

# Create a prompt using the Agent's configuration
prompt = f"""
You are {agent.role}.

Your goal is:
{agent.goal}

Background:
{agent.backstory}

Question:
{question}

Explain the answer in simple and clear language and in  lines.
"""

# Use the underlying LLM separately
response = agent.llm.invoke(prompt)

# Display the answer
print("\nAnswer:")
print(response.content)
```

**AGENT WITH CONTEXT WINDOW** - *respect_context_window=True tells the agent to manage information so it stays within the LLM’s available context window.*

```python
from crewai import Agent

agent = Agent(
    role="AI Teacher",
    goal="Explain AI concepts clearly",
    backstory="You are an experienced AI teacher.",
    respect_context_window=True,  
    verbose=True
)
question = input("Enter your question: ")
result = await agent.kickoff(question)
print("\nAnswer:")
print(result)
```

**AGENT USING TOOL**

```python
from crewai import Agent
from crewai_tools import SerperDevTool

search_tool = SerperDevTool()

agent = Agent(
    role="Web Researcher",
    goal="Find and explain information from websites",
    backstory="You are a researcher who searches websites and provides useful answers.",
    tools=[search_tool],
    verbose=True
)

question = input("Enter your question: ")
result = await agent.kickoff(question)

print("\nAnswer:")
print(result)
```

**AGENT USING KNOWLEDGE SOURCE**

```python
from crewai import Agent
from crewai.knowledge.source.pdf_knowledge_source import PDFKnowledgeSource

knowledge_source = PDFKnowledgeSource(
    file_paths=["Hospital_Policy.pdf"]
)

agent = Agent(
    role="Hospital Assistant",
    goal="Answer questions using company policy",
    backstory="You are an Hospital assistant who helps employees understand the rules.",
    knowledge_sources=[knowledge_source],
    verbose=True
)

question = input("Enter your question: ")
result = await agent.kickoff(question)

print("\nAnswer:")
print(result)
```
