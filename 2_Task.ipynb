**Task is a specific piece of work that you assign to an Agent.**

**Task tells the agent three main things:**<br>
-what to do<br>
-what result is expected<br>
-which agent should do it

**The most commonly used Task properties:**
| Property          | Purpose                                        |
| ----------------- | ---------------------------------------------- |
| `description`     | Explains what the agent should do              |
| `expected_output` | Defines what the final result should look like |
| `agent`           | Specifies which agent performs the task        |
| `tools`           | Tools available specifically for the task      |
| `context`         | Output from other tasks that this task can use |
| `async_execution` | Allows a task to execute asynchronously        |
| `output_file`     | Saves the task result to a file                |

**SIMPLE AGENT+TASK+CREW**

**Running a Crew with kickoff()**<br>
**kickoff() is a CrewAI method used to start the crew’s execution.**<br>

**It is used to:**<br>
-Pass input values to tasks.<br>
-Make agents perform their assigned tasks.<br>
-Execute tasks using the selected process.<br>
-Return the crew’s final output.<br>

```python
from crewai import Agent, Task, Crew

agent = Agent(
    role="AI Teacher",
    goal="Explain AI concepts clearly",
    backstory="You are an experienced AI teacher.",
    verbose=True
)

task = Task(
    description="Explain Generative AI in simple terms.",
    expected_output="A simple explanation of Generative AI.",
    agent=agent
)

crew = Crew(
    agents=[agent],
    tasks=[task],
    verbose=True
)

result = await crew.kickoff_async()

print("\nAnswer:")
print(result)
```

**AGENT + TASK + CREW WITH TOOL AND GUARDRAILS**

```python
from crewai import Agent, Task, Crew
from crewai_tools import SerperDevTool

# Search Tool
search_tool = SerperDevTool()

# Guardrail Function
def validate_answer(result):
    if len(result.raw.split()) <= 100:
        return True, result.raw
    else:
        return False, "Answer must be under 100 words."


# Agent
agent = Agent(
    role="AI Teacher",
    goal="Explain AI concepts clearly using reliable information",
    backstory="You are an experienced AI teacher.",
    verbose=True
)

# Task
task = Task(
    description="Search for information about Generative AI and explain Generative AI in simple terms.", 
    expected_output="A beginner-friendly explanation of Generative AI under 200 words.",
    agent=agent,
    tools=[search_tool],
    guardrail=validate_answer
)

# Crew
crew = Crew(
    agents=[agent],
    tasks=[task],
    verbose=True
)

# Run Crew
result = await crew.kickoff_async()

print("\nAnswer:")
print(result)
```

**2-AGENTS AND 2-TASKS WITH PROCESS**

**Sequential: "First Agent 1 works, then Agent 2 works."**<br>
**Hierarchical: "The Manager decides how the Agents should work." [In a hierarchical Crew, CrewAI requires a manager agent or manager LLM.]**<br>

```python
from crewai import Agent, Task, Crew, Process

# Agent 1
research_agent = Agent(
    role="AI Researcher",
    goal="Collect simple and accurate information about AI topics",
    backstory="You are an AI researcher who explains technical concepts clearly.",
    verbose=True
)

# Agent 2
teacher_agent = Agent(
    role="AI Teacher",
    goal="Teach AI concepts in a beginner-friendly way",
    backstory="You are an experienced AI teacher who makes complex topics easy to understand.",
    verbose=True
)

# Task 1
research_task = Task(
    description="Find the important points about Generative AI.",
    expected_output="A short list of important points about Generative AI.",
    agent=research_agent
)

# Task 2
teaching_task = Task(
    description="Explain Generative AI in simple terms for beginners.",
    expected_output="A simple beginner-friendly explanation of Generative AI.",
    agent=teacher_agent
)

# Crew
crew = Crew(
    agents=[research_agent, teacher_agent],
    tasks=[research_task, teaching_task],
    process=Process.hierarchical,
    manager_llm="gpt-4o-mini",
    verbose=True
)

# Run Crew
result = await crew.kickoff_async()

print("\nFinal Answer:")
print(result)
```

**EXAMPLE DEFINING CLASS AND FUNCTION**

```python
from crewai import Agent, Task, Crew, Process

class AITeacherCrew:

    def create_crew(self):

        # Agent
        teacher_agent = Agent(
            role="AI Teacher",
            goal="Explain AI concepts clearly",
            backstory="You are an experienced AI teacher who teaches beginners.",
            verbose=True
        )

        # Task
        ai_task = Task(
            description="Explain Generative AI in simple terms.",
            expected_output="A simple beginner-friendly explanation of Generative AI.",
            agent=teacher_agent
        )

        # Crew
        crew = Crew(
            agents=[teacher_agent],
            tasks=[ai_task],
            process=Process.sequential,
            verbose=True
        )

        return crew

# Async Function
async def run_crew():

    ai_crew = AITeacherCrew()
    crew = ai_crew.create_crew()
    result = await crew.kickoff_async()
    return result


# Run in Jupyter Notebook
output = await run_crew()

print("\nFinal Answer:")
print(output)
```
