**STATE MANAGEMENT - State management in CrewAI means storing, updating, and sharing data between steps of a workflow. In CrewAI Flows, this data is accessed using self.state**

| Type                   | How data is stored                                | 
| ---------------------- | ------------------------------------------------- |
| **Unstructured state** | A dictionary with flexible keys                   | 
| **Structured state**   | A Pydantic model with predefined fields and types | 


**Unstructured state - Stores data in a dictionary. Access values using self.state["key"].**

```python
from crewai.flow.flow import Flow, start, listen

class UnstructuredFlow(Flow):

    @start()
    def set_score(self):
        self.state["score"] = 75

    @listen(set_score)
    def show_result(self):
        self.state["result"] = (
            "PASS" if self.state["score"] >= 50 else "FAIL"
        )
        print("Score:", self.state["score"])
        print("Result:", self.state["result"])

flow = UnstructuredFlow()
flow.plot()
await flow.kickoff_async()
```

**Structured state - Defines fields and their types using a Pydantic model. Access values using self.state.field.**

```python
from pydantic import BaseModel
from crewai.flow.flow import Flow, start, listen

class StudentState(BaseModel):
    score: int = 0
    result: str = ""

class StructuredFlow(Flow[StudentState]):

    @start()
    def set_score(self):
        self.state.score = 75

    @listen(set_score)
    def show_result(self):
        self.state.result = (
            "PASS" if self.state.score >= 50 else "FAIL"
        )
        print("Score:", self.state.score)
        print("Result:", self.state.result)

flow = StructuredFlow()
flow.plot()
await flow.kickoff_async()
```

**Updating state means changing a value already stored in self.state.**

```python
from crewai.flow.flow import Flow, start, listen

class CounterFlow(Flow):

    @start()
    def set_counter(self):
        self.state["counter"] = 0
        print("Before update:", self.state["counter"])

    @listen(set_counter)
    def update_counter(self):
        self.state["counter"] += 1
        print("After update:", self.state["counter"])

flow = CounterFlow()
flow.plot()
await flow.kickoff_async()
```
