**SEQUENTIAL PROCESS - Executes tasks one after another in a fixed order. The output of an earlier task can be used by the next task.**

**HIERARCHICAL PROCESS - It is a workflow where a manager agent coordinates and delegates tasks to other agents based on their roles and abilities.**

**SEQUENTIAL PROCESS**

```python
from crewai import Agent, Task, Crew, Process

# Agent 1
researcher = Agent(
    role="AI Researcher",
    goal="Research the user's question",
    backstory="You collect useful and accurate information.",
    verbose=True
)

# Agent 2
writer = Agent(
    role="AI Writer",
    goal="Write a simple beginner-friendly answer",
    backstory="You convert research into clear and simple explanations.",
    verbose=True
)

# Task 1
research_task = Task(
    description="Research this question: {question}",
    expected_output="Important points related to the user's question.",
    agent=researcher
)

# Task 2
writing_task = Task(
    description="Using the research, give a simple answer for: {question}",
    expected_output="A clear beginner-friendly final answer.",
    agent=writer
)

# Crew
crew = Crew(
    agents=[researcher, writer],    
    tasks=[research_task, writing_task],
    process=Process.sequential,
    verbose=True
)

# Run
result = await crew.kickoff_async(inputs={"question": "What are the applications of Artificial Intelligence?"})

print("\nFinal Answer:")
print(result)
```

**HIERARCHICAL PROCESS**

```python
from crewai import Agent, Task, Crew, Process, LLM

#llm = LLM(model="gpt-4o-mini", temperature=0)

# Agent 1
researcher = Agent(
    role="AI Researcher",
    goal="Research the user's question",
    backstory="You collect useful and accurate information.",
    verbose=True
)

# Agent 2
writer = Agent(
    role="AI Writer",
    goal="Write a simple beginner-friendly answer",
    backstory="You convert research into clear and simple explanations.",
    verbose=True
)

# Task 1: The manager chooses an agent to do the research
research_task = Task(
    description="Research this question: {question}",
    expected_output="Important points related to the user's question."
)

# Task 2: Use the research output to write the answer
writing_task = Task(
    description="Using the research, give a simple answer for: {question}",
    expected_output="A clear beginner-friendly final answer.",
    context=[research_task]
)

# Crew with a manager
crew = Crew(
    agents=[researcher, writer],
    tasks=[research_task, writing_task],
    process=Process.hierarchical,
    manager_llm="gpt-4o-mini",
    verbose=True
)

result = await crew.kickoff_async(inputs={"question": "What are the applications of Artificial Intelligence?"})

print("\nFinal Answer:")
print(result)
```
